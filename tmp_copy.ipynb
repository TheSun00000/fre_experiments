```python
import os
import shutil
```

```python
# Paths to source (A) and destination (B)
source_folder = "/lustre/fswork/projects/rech/vcj/uyo96iu/fre_experiments/models/2025-10-28_21-30-43_FB-antmaze"
destination_folder = "/lustre/fswork/projects/rech/vcj/uyo96iu/fre_experiments/shared_models/antmaze_test"

# Create destination folder if it doesn't exist
os.makedirs(destination_folder, exist_ok=True)

# Iterate over all files in source folder
for filename in os.listdir(source_folder):
    src_path = os.path.join(source_folder, filename)
    dst_path = os.path.join(destination_folder, filename)
    
    id = int(src_path.split('.')[0].split(':')[-1])
    
    if id == 5000 or (id <= 200_000 and id % 10_000 == 0):
        print(id)

        # Only copy files (ignore directories)
        if os.path.isfile(src_path):
            shutil.copy2(src_path, dst_path)  # copy2 preserves metadata

print("Files copied successfully!")
```
